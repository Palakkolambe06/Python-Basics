```python
# Program 3: College Attendance Management
students = ["Amit", "Priya", "Rahul", "Neha", "Riya"]
present_count = 0

# Check attendance of each student
for student in students:
    status = input("Is " + student + " present? (yes/no): ")
    if status.lower() == "yes":
        present_count += 1
print("Total Students:", len(students))
print("Present Students:", present_count)
print("Absent Students:", len(students) - present_count)
```
