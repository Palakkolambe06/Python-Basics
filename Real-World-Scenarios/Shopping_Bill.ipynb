```python
# Program 2: Shopping Bill Calculator
total_bill = 0
products = ["Shoes", "Bag", "Watch", "T-Shirt"]
prices = [1500, 800, 1200, 600]

# Calculate the total price of all products
for i in range(len(products)):
    print(products[i], "=", prices[i])
    total_bill += prices[i]

print("Total Shopping Bill: Rs.", total_bill)
```
