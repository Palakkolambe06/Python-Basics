```python
# Program 4: Restaurant Food Ordering System
total = 0
while True:
    print("\n1. Pizza - Rs. 200")
    print("2. Burger - Rs. 100")
    print("3. Sandwich - Rs. 80")
    print("4. Exit")
    choice = int(input("Enter your choice: "))

    if choice == 1:
        total += 200
        print("Pizza added to order.")

    elif choice == 2:
        total += 100
        print("Burger added to order.")

    elif choice == 3:
        total += 80
        print("Sandwich added to order.")

    elif choice == 4:
        break

    else:
       print("Invalid choice!")
print("Your Total Bill: Rs.", total)
```
