```python
# Program 1: ATM PIN Verification System
correct_pin = "1234"
attempts = 0
while attempts < 3:
    pin = input("Enter your ATM PIN: ")
    if pin == correct_pin:
        print("Login Successful! Welcome.")
        break
    else:
        attempts += 1
        print("Incorrect PIN. Attempts left:", 3 - attempts)
if attempts == 3:
    print("Your ATM card is temporarily blocked.")
```
