```python
# Using else block
try:
    num = int(input("Enter Number: "))
    result = 100 / num

except ZeroDivisionError:
    print("Division by Zero")

else:
    # Executes only when no exception occurs
    print("Result:", result)
```

```python
# Using else and finally together
try:
    num = int(input("Enter the number: "))
    result = 50 / num

except ZeroDivisionError:
    print("Cannot divide by zero")

except ValueError:
    print("Use Integer only")

else:
    # Executes when no exception occurs
    print("Result:", result)

finally:
    # Always executes
    print("Thankyou for using the program")
```

```python
# Using Exception Object
try:
    a = 10 / 0

except ZeroDivisionError as e:
    # 'e' stores the exception message
    print("Exception Message:", e)
```

```python
# Using finally block
try:
    print("Processing Transaction")
    amount = int(input("Enter amount: "))

    if amount > 10000:
        print("Transaction Approved")

except:
    # Executes if an exception occurs
    print("Transaction Failed")

finally:
    # Always executes
    print("Thankyou")
```

```python
# Handling multiple exceptions in one block
try:
    num = int(input("Enter Number: "))
    result = 100 / num
    print(result)

except (ValueError, ZeroDivisionError):
    # Handles both invalid input and division by zero
    print("Invalid Operation")
```

```python
# Using raise keyword for password validation
try:
    password = input("Enter the password: ")

    if len(password) < 8:
        raise ValueError("Password must contain at least 8 characters.")

    if not password.isalpha():
        # isalpha() checks whether all characters are letters
        raise ValueError("Password must contain letters only.")

    print("Valid Password")

except ValueError as e:
    # Displays the custom exception message
    print(e)
```

```python
# Using multiple except blocks
try:
    num = int(input("Enter Number: "))
    print(100 / num)

except ZeroDivisionError:
    print("Division by Zero")

except ValueError:
    print("Invalid Input")
```

```python

# ATM Withdrawal using Exception Handling
try:
    balance = 5000

    amount = int(input("Enter withdrawal amount: "))
    if amount <= 0:
        raise ValueError("Amount must be positive")
    if amount > balance:
        raise Exception("Insufficient Balance")

    balance = balance - amount
except ValueError as e:
    print("Invalid Input:", e)

except Exception as e:
    print("Transaction Failed:", e)
else:
    print("Transaction Successful!")
    print("Remaining Balance:", balance)
finally:
    print("Thank you for using ATM!")
```
